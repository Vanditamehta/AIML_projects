```python

!pip install PyPDF2

# ==========================================
# IMPORT LIBRARIES
# ==========================================
import os
import re
import numpy as np
import PyPDF2

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

# ==========================================
# SYSTEM HEADER
# ==========================================
print("\n" + "="*65)
print("        SMART RESUME ANALYZER & JOB MATCHER")
print("="*65)

# ==========================================
# STEP 1: LOAD RESUME
# ==========================================
file_name = input("\nEnter resume file name (PDF/TXT): ")

if not os.path.exists(file_name):
    print(" File not found. Upload in Files tab.")
    exit()

resume_text = ""

if file_name.endswith(".pdf"):
    with open(file_name, "rb") as f:
        reader = PyPDF2.PdfReader(f)
        for page in reader.pages:
            text = page.extract_text()
            if text:
                resume_text += text

elif file_name.endswith(".txt"):
    with open(file_name, "r") as f:
        resume_text = f.read()

else:
    print(" Unsupported format")
    exit()

print(" Resume Loaded Successfully")

# ==========================================
# STEP 2: SELECT JOB ROLE
# ==========================================
print("\nSelect Job Role:")
print("1. Software Developer")
print("2. Data Scientist")
print("3. Web Developer")
print("4. AI Engineer")

choice = int(input("Enter choice (1-4): "))

job_roles = {
    1: "python java c sql problem solving teamwork",
    2: "python machine learning data science ai deep learning",
    3: "html css javascript communication teamwork web design",
    4: "python ai deep learning machine learning neural networks"
}

role_names = {
    1: "Software Developer",
    2: "Data Scientist",
    3: "Web Developer",
    4: "AI Engineer"
}

job_desc = job_roles.get(choice, "")
selected_role = role_names.get(choice, "Unknown")

print(f"\n Selected Role: {selected_role}")

# ==========================================
# STEP 3: TEXT CLEANING
# ==========================================
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    return text

resume_text = clean_text(resume_text)
job_desc = clean_text(job_desc)

# ==========================================
# STEP 4: KEYWORDS
# ==========================================
keywords = [
    "python", "java", "c", "machine learning",
    "data science", "sql", "html", "css",
    "javascript", "deep learning", "ai",
    "communication", "teamwork", "problem solving"
]

# ==========================================
# STEP 5: FEATURE EXTRACTION
# ==========================================
def extract_features(text):
    return np.array([1 if word in text else 0 for word in keywords])

resume_features = extract_features(resume_text)
job_features = extract_features(job_desc)

# Debug
print("\n Feature Check:")
print("Resume:", resume_features)
print("Job   :", job_features)

X = np.array([resume_features, job_features])

# ==========================================
# STEP 6: FEATURE SCALING
# ==========================================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ==========================================
# STEP 7: PCA
# ==========================================
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# ==========================================
# STEP 8: SYNTHETIC TRAINING DATA
# ==========================================
np.random.seed(42)
train_X = np.random.rand(100, 2)
train_y = np.random.randint(0, 2, 100)

# ==========================================
# STEP 9: MODELS
# ==========================================
svm = SVC(probability=True)
svm.fit(train_X, train_y)

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(train_X, train_y)

tree = DecisionTreeClassifier()
tree.fit(train_X, train_y)

# ==========================================
# STEP 10: PREDICTION
# ==========================================
input_data = X_pca[0].reshape(1, -1)

svm_pred = svm.predict(input_data)
knn_pred = knn.predict(input_data)
tree_pred = tree.predict(input_data)

confidence = svm.predict_proba(input_data).max()

# ==========================================
# STEP 11: MATCH SCORE
# ==========================================
dot = np.dot(resume_features, job_features)
norm = (np.linalg.norm(resume_features) * np.linalg.norm(job_features))

match_score = 0 if norm == 0 else (dot / norm)
match_score = round(match_score * 100, 2)

# ==========================================
# STEP 12: OUTPUT REPORT
# ==========================================
print("\n" + "="*65)
print("                RESUME ANALYSIS REPORT")
print("="*65)

print(f"\n Target Role              : {selected_role}")
print(f" Resume Match Score       : {match_score}%")
print(f" Model Confidence         : {round(confidence*100,2)}%")

print("\n AI Evaluation")
print("-"*65)

score_sum = svm_pred[0] + knn_pred[0] + tree_pred[0]

if score_sum <= 1:
    print("Candidate Profile           : Strong Alignment")
elif score_sum == 2:
    print("Candidate Profile           : Moderate Alignment")
else:
    print("Candidate Profile           : Low Alignment")

print("\n Insights")
print("-"*65)

if match_score > 70:
    print("• Resume strongly matches job requirements")
elif match_score > 40:
    print("• Resume partially matches job role")
else:
    print("• Resume needs improvement for this role")

print("\n Missing Skills")
print("-"*65)

missing = [keywords[i] for i in range(len(keywords))
           if resume_features[i] == 0 and job_features[i] == 1]

if missing:
    print("• Consider adding:", ", ".join(missing))
else:
    print("• All key skills are present in your resume")

print("\n DISCLAIMER")
print("-"*65)
print("This system provides AI-assisted resume analysis.")
print("Results are indicative and not final hiring decisions.")

print("="*65)
```
